```python
import google.generativeai as genai
```

```python
from google.colab import userdata
passkey = userdata.get('GOOGLE_API_KEY')
```

```python
genai.configure(api_key=passkey)
```

```python
from PIL import Image as image
img =image.open("/content/ask.png")
```

```python
model= genai.GenerativeModel("gemini-2.5-flash")
```

```python
from time import sleep
```

```python
while True:
  prompt=input('user:')
  print("-"*100)
  if prompt.lower() not in ["bye","exit","quit"]:
    response=model.generate_content([prompt,img])
    print("Bot",response.text)
    print()
    print("-"*100)
    print()
  else:
    print("exitting.........")
    sleep(3)
    print("see you")
    break
```
